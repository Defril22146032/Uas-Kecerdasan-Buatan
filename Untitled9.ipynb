```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Agar hasil random konsisten
random.seed(42)
np.random.seed(42)
```

```python
# Data Mata Kuliah dan Dosen
courses = [
    ("AI", "DosenA"),
    ("ML", "DosenB"),
    ("DB", "DosenC"),
    ("CN", "DosenD"),
    ("SE", "DosenE"),
    ("OS", "DosenF"),
    ("IR", "DosenA"),
    ("NLP", "DosenB"),
    ("CV", "DosenC"),
    ("IOT", "DosenD"),
    ("HCI", "DosenE"),
    ("CyberSec", "DosenF"),
    ("BigData", "DosenA"),
    ("DL", "DosenB"),
    ("NoSQL", "DosenC"),
    ("Cloud", "DosenD"),
    ("UX", "DosenE"),
    ("SecEng", "DosenF"),
    ("Agile", "DosenA"),
    ("RL", "DosenB"),
    ("ETL", "DosenC"),
    ("DevOps", "DosenD"),
    ("GameDev", "DosenE"),
    ("Blockchain", "DosenF")
]

# Data ruang
rooms = ["R1", "R2", "R3", "R4"]

# Data slot waktu
timeslots = [
    "Senin-08",
    "Senin-10",
    "Senin-13",
    "Selasa-08",
    "Selasa-10",
    "Selasa-13",
    "Rabu-08",
    "Rabu-10",
    "Rabu-13",
    "Kamis-08",
    "Kamis-10",
    "Kamis-13"
]

print("Jumlah Mata Kuliah :", len(courses))
print("Jumlah Dosen       :", len(set(dosen for _, dosen in courses)))
print("Jumlah Ruang       :", len(rooms))
print("Jumlah Slot Waktu  :", len(timeslots))
```

```python
POPULATION_SIZE = 60
MAX_GENERATIONS = 100

CROSSOVER_RATE = 0.85
MUTATION_RATE = 0.20

ELITISM_COUNT = 2
```

```python
def create_individual():
    """
    Membuat satu individu/jadwal yang terdiri dari 24 gen.
    Setiap gen menentukan ruang dan slot waktu.
    """

    individual = []

    for _ in courses:
        room = random.randrange(len(rooms))
        timeslot = random.randrange(len(timeslots))

        individual.append((room, timeslot))

    return individual
```

```python
def create_population():
    """
    Membuat populasi awal sebanyak 60 individu.
    """

    return [
        create_individual()
        for _ in range(POPULATION_SIZE)
    ]


population = create_population()

print("Jumlah individu dalam populasi:", len(population))
print("Jumlah gen setiap individu:", len(population[0]))
```

```python
def calculate_conflicts(individual):
    """
    Menghitung jumlah konflik pada sebuah jadwal.
    """

    room_conflicts = 0
    lecturer_time_conflicts = 0
    lecturer_day_conflicts = 0

    # Membandingkan setiap pasangan mata kuliah
    for i in range(len(courses)):

        course_i, lecturer_i = courses[i]
        room_i, time_i = individual[i]

        # Hari diambil dari nama timeslot
        day_i = timeslots[time_i].split("-")[0]

        for j in range(i + 1, len(courses)):

            course_j, lecturer_j = courses[j]
            room_j, time_j = individual[j]

            day_j = timeslots[time_j].split("-")[0]

            # --------------------------------
            # 1. Konflik Ruang
            # --------------------------------
            if room_i == room_j and time_i == time_j:
                room_conflicts += 1

            # --------------------------------
            # 2. Konflik Dosen-Waktu
            # --------------------------------
            if lecturer_i == lecturer_j and time_i == time_j:
                lecturer_time_conflicts += 1

            # --------------------------------
            # 3. Konflik Dosen-Hari
            # --------------------------------
            if lecturer_i == lecturer_j and day_i == day_j:
                lecturer_day_conflicts += 1

    total_conflicts = (
        room_conflicts +
        lecturer_time_conflicts +
        lecturer_day_conflicts
    )

    return (
        total_conflicts,
        room_conflicts,
        lecturer_time_conflicts,
        lecturer_day_conflicts
    )
```

```python
def fitness(individual):
    """
    Menghasilkan nilai fitness 0-100.
    Semakin sedikit konflik, semakin tinggi fitness.
    """

    total_conflicts, _, _, _ = calculate_conflicts(individual)

    score = 100 / (1 + total_conflicts)

    return score
```

```python
def selection(population, tournament_size=3):
    """
    Memilih satu individu menggunakan Tournament Selection.
    """

    participants = random.sample(population, tournament_size)

    winner = max(
        participants,
        key=fitness
    )

    return winner.copy()
```

```python
def crossover(parent1, parent2):
    """
    One-point crossover.
    """

    # Tidak melakukan crossover jika random > 85%
    if random.random() > CROSSOVER_RATE:
        return parent1.copy(), parent2.copy()

    # Menentukan titik potong
    point = random.randint(1, len(parent1) - 1)

    child1 = (
        parent1[:point] +
        parent2[point:]
    )

    child2 = (
        parent2[:point] +
        parent1[point:]
    )

    return child1, child2
```

```python
def mutation(individual):
    """
    Mutasi gen dengan probabilitas 20%.
    """

    mutated = individual.copy()

    for i in range(len(mutated)):

        if random.random() < MUTATION_RATE:

            # Pilih ruang dan waktu baru
            new_room = random.randrange(len(rooms))
            new_timeslot = random.randrange(len(timeslots))

            mutated[i] = (new_room, new_timeslot)

    return mutated
```

```python
def get_elites(population):
    """
    Mengambil individu terbaik sebanyak ELITISM_COUNT.
    """

    sorted_population = sorted(
        population,
        key=fitness,
        reverse=True
    )

    return [
        individual.copy()
        for individual in sorted_population[:ELITISM_COUNT]
    ]
```

```python
def create_new_population(population):
    """
    Membentuk populasi generasi berikutnya.
    """

    new_population = []

    # --------------------------------
    # Elitism
    # --------------------------------
    elites = get_elites(population)
    new_population.extend(elites)

    # --------------------------------
    # Crossover + Mutation
    # --------------------------------
    while len(new_population) < POPULATION_SIZE:

        parent1 = selection(population)
        parent2 = selection(population)

        child1, child2 = crossover(
            parent1,
            parent2
        )

        child1 = mutation(child1)
        child2 = mutation(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:
            new_population.append(child2)

    return new_population
```

```python
# Membuat populasi awal
population = create_population()

# Menyimpan histori fitness
best_fitness_history = []
average_fitness_history = []

# Menyimpan individu terbaik
best_individual = None
best_fitness_value = -1

# Menjalankan 100 generasi
for generation in range(MAX_GENERATIONS):

    # Fitness seluruh individu
    fitness_values = [
        fitness(individual)
        for individual in population
    ]

    # Fitness maksimum
    max_fitness = max(fitness_values)

    # Fitness rata-rata
    avg_fitness = np.mean(fitness_values)

    # Simpan histori
    best_fitness_history.append(max_fitness)
    average_fitness_history.append(avg_fitness)

    # Update individu terbaik
    max_index = np.argmax(fitness_values)

    if max_fitness > best_fitness_value:
        best_fitness_value = max_fitness
        best_individual = population[max_index].copy()

    # Tampilkan perkembangan
    print(
        f"Generasi {generation + 1:3d} | "
        f"Fitness Maks: {max_fitness:6.2f} | "
        f"Fitness Rata-rata: {avg_fitness:6.2f}"
    )

    # Jika sudah tidak ada konflik
    if calculate_conflicts(best_individual)[0] == 0:
        print("\nSolusi tanpa konflik ditemukan!")
        break

    # Membuat generasi berikutnya
    population = create_new_population(population)
```

```python
plt.figure(figsize=(12, 6))

generations = range(
    1,
    len(best_fitness_history) + 1
)

plt.plot(
    generations,
    best_fitness_history,
    label="Fitness Maksimum"
)

plt.plot(
    generations,
    average_fitness_history,
    label="Fitness Rata-rata"
)

plt.title("Evolusi Fitness Algoritma Genetika")
plt.xlabel("Generasi")
plt.ylabel("Fitness")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
def create_schedule_dataframe(individual):
    """
    Mengubah individu terbaik menjadi DataFrame jadwal.
    """

    schedule = []

    for i, (course, lecturer) in enumerate(courses):

        room_index, timeslot_index = individual[i]

        timeslot = timeslots[timeslot_index]
        day, time = timeslot.split("-")

        schedule.append({
            "No": i + 1,
            "Mata Kuliah": course,
            "Dosen": lecturer,
            "Hari": day,
            "Jam": time,
            "Ruang": rooms[room_index]
        })

    return pd.DataFrame(schedule)


schedule_df = create_schedule_dataframe(
    best_individual
)

schedule_df
```
